## 3. Visualización de precios
Grafica en una misma figura la evolución de los precios de cierre ajustados de las tres acciones.

🔎 **Pregunta:** ¿Cuál de las tres acciones muestra mayor crecimiento en estos 3 años?

```python
import yfinance as yf
import matplotlib.pyplot as plt

# Descargar datos de los últimos 5 años
acciones = ["GOOG", "AMZN", "MSFT"]
datos = yf.download(tickers=acciones, period="5y")["Close"]

# Calcular promedio móvil de 30 días para cada acción
MA30 = datos.rolling(window=30).mean()

# Graficar todas en una sola figura
plt.figure(figsize=(12,6))
for ticker in acciones:
    plt.plot(datos.index, datos[ticker], label=f"{ticker} Precio")
    plt.plot(MA30.index, MA30[ticker], linestyle="--", label=f"{ticker} MA30")

plt.title("Acciones GOOG, AMZN y MSFT con promedio móvil de 30 días")
plt.xlabel("Fecha")
plt.ylabel("Precio (USD)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Calcular el crecimiento porcentual de cada acción
crec_msft = (datos["MSFT"].iloc[-1] / datos["MSFT"].iloc[0] - 1) * 100
crec_goog = (datos["GOOG"].iloc[-1] / datos["GOOG"].iloc[0] - 1) * 100
crec_amzn = (datos["AMZN"].iloc[-1] / datos["AMZN"].iloc[0] - 1) * 100

# Mostrar crecimientos
print("Crecimientos (%):")

print(f"MSFT: {crec_msft:.2f}%")
print(f"GOOG: {crec_goog:.2f}%")
print(f"AMZN: {crec_amzn:.2f}%")

if crec_msft > crec_goog and crec_msft > crec_amzn:
    print("Microsoft tuvo el mayor crecimiento")
elif crec_goog > crec_msft and crec_goog > crec_amzn:
    print("Google tuvo el mayor crecimiento")
elif crec_amzn > crec_goog and crec_amzn > crec_msft:
    print("Amazon tuvo el mayor crecimiento")
else:
    print("Hay un empate en el crecimiento")
```
